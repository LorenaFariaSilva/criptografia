```python
dicionario = "qwertyuiopasdfghjklzxcvbnm"
alfabeto = "abcdefghijklmnopqrstuvwxyz"

def codifica(senha):
  senha_criptografada = ""
  for letra in senha:
    if "a" <= letra <= "z":
       posicao = ord(letra) - ord("a") # 0 até 25
       senha_criptografada += dicionario[posicao]
    else:
     senha_criptografada += letra
  return senha_criptografada

senha = "Marcelo123"
print(f"A senha '{senha}' criptografada é: {codifica(senha)}")



# Exemplo de uso com uma nova senha:
nova_senha = "python"
print(f"A senha '{nova_senha}' criptografada é: {codifica(nova_senha)}")
```

```python
def calcula_hash(senha):
  valor = 0
  for letra in senha:
     valor = valor*31 + ord(letra)
     valor %= 2**32
  return f"0x{valor:08x}"


print(calcula_hash("Marcelo123!@"))
print(calcula_hash("Ma"))
print(calcula_hash("Marcelo-criptografia-hash-cifras-bilhete-colab-gemini"))
```

```python
def calcula_hash(senha):
  valor = 0
  for letra in senha:
     valor = valor*31 + ord(letra)
     valor %= 2**32
  return f"0x{valor:08x}"


# print(calcula_hash("Marcelo123!@"))
# print(calcula_hash("Ma"))
# print(calcula_hash("Marcelo-criptografia-hash-cifras-bilhete-colab-gemini"))

senha_cadastrada = "marcelo123"
hash_cadastrado = calcula_hash(senha_cadastrada)

senha_digitada = input("Digite sua senha : ")
hash_digitado = calcula_hash(senha_digitada)

if hash_cadastrado == hash_digitado:
    print("ACESSO LIBERADO")
else:
    print("ACESSO NEGADO")
```

```python
senhas_comuns = [
    "123456",
    "password",
    "123456789",
    "qwerty",
    "12345678",
    "111111",
    "12345",
    "columbia",
    "senha",
    "iloveyou"
]

hashs_comuns = []
for senha in senhas_comuns:
    hashs_comuns.append(calcula_hash(senha))

print(hashs_comuns)
```

```python
import itertools

combinacoes = itertools.product([hash_cadastrado], hashs_comuns)
print(list(combinacoes))
```

```python
senha_nova = "mar"
hash_novo = calcula_hash(senha_nova)

dicionario_novo = "abcdefghijklmnopqrstuvwxyzABC123!@#$"

for combinacao in itertools.product(dicionario_novo, repeat=len(senha_nova)):
  if calcula_hash(combinacao) == hash_novo:
    print(combinacao)
    break
```
